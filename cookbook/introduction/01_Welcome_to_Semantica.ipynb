Semantica 由三个概念层和多个具体模块组成。

### 分层

- **输入层（Input Layer）**
  - 连接文件、网页、API、数据库、邮件、订阅源、代码仓库和数据流

  - 与向量存储、图数据库和下游应用集成

## 🧩 Semantica 模块参考

Semantica 采用模块化设计。以下是按功能分组的所有可用模块的完整指南。

### 📥 接入与解析
处理原始数据输入和结构化的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`ingest`** | **数据接入**<br>连接各类数据源。 | • 文件、网页、订阅源、数据流接入<br>• 支持数据库、邮件、代码仓库、MCP |
| **`parse`** | **文档解析**<br>将原始内容解析为结构化数据。 | • PDF、HTML、JSON、CSV、Excel<br>• 图片与代码解析 |

### ⚙️ 数据处理
清洗、规范化和切分数据的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`normalize`** | **数据规范化**<br>清洗并标准化文本。 | • 文本清洗与语言检测<br>• 实体、日期、数字规范化 |
| **`split`** | **文本切分**<br>为 RAG 切分文档。 | • 递归字符切分<br>• 语义切分与按 Token 切分 |

### 🧠 抽取与增强
从原始数据中抽取含义、结构和向量的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`semantic_extract`** | **信息抽取**<br>抽取实体和关系。 | • NER 与关系抽取<br>• 事件与语义网络检测 |
| **`context`** | **智能体记忆**<br>管理 AI 智能体的状态。 | • 长期记忆与历史记录<br>• 上下文图与 RAG 集成 |

### 🕸️ 知识图谱核心
构建、精炼和消解知识图谱的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`kg`** | **图构建**<br>构建和分析图谱。 | • 图构建与图分析<br>• 校验与实体消解 |
| **`conflicts`** | **冲突消解**<br>解决数据矛盾。 | • 来源可靠性评分<br>• 真值发现算法 |
| **`deduplication`** | **实体消解**<br>合并重复实体。 | • 基于相似度的分块<br>• 聚类与规范化归并 |

### 💾 存储与检索
持久化和查询数据的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`embeddings`** | **向量 Embedding**<br>生成语义向量。 | • 文本与图 Embedding<br>• 多提供商支持（OpenAI 等） |
| **`vector_store`** | **向量数据库**<br>存储和检索向量。 | • 相似度搜索与过滤<br>• 混合搜索（向量 + 关键词） |
| **`graph_store`** | **属性图存储**<br>持久化图数据。 | • Neo4j、FalkorDB 适配器<br>• Cypher 查询支持 |
| **`triplet_store`** | **RDF 存储**<br>持久化语义三元组。 | • SPARQL 端点<br>• BlazeGraph、Jena、Virtuoso 适配器 |

### 🔎 推理与分析
推导新知识并评估质量的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`reasoning`** | **推理器门面**<br>统一的推理接口。 | • Datalog/基于规则的推理<br>• 前向/后向链式推理 |
| **`ontology`** | **本体管理**<br>管理 schema 和定义。 | • 从数据生成本体<br>• 验证与演化 |
| **`visualization`** | **可视化分析**<br>可视化图谱与指标。 | • 2D/3D 图可视化<br>• 交互式图表与仪表盘 |
| **`evals`** | **评估**<br>对流水线质量做基准测试。 | • RAG 与图质量指标<br>• 与标准答案对比 |

### 🛠️ 编排与工具
管理框架和工作流的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`core`** | **框架核心**<br>主入口与配置。 | • 生命周期管理<br>• 插件系统与配置 |
| **`pipeline`** | **工作流编排**<br>管理复杂流程。 | • DAG 执行与重试<br>• 错误处理与可观测性 |
| **`seed`** | **数据播种**<br>初始化知识库。 | • 分类体系与本体播种<br>• 参考数据加载 |
| **`export`** | **数据导出**<br>将数据导出为文件。 | • JSON、CSV、RDF、GEXF 导出<br>• 报告生成 |
| **`utils`** | **工具集**<br>通用辅助函数。 | • 日志、异步、哈希<br>• 文本处理辅助函数 |

## 核心概念（高层视角）

- **知识图谱**
  - 节点表示实体，如人物、组织、地点、事件或概念
  - 边表示关系，如 `works_for`、`located_in`、`founded_by`
  - 属性承载特性和元数据，如时间戳、来源和置信度

- **实体与关系**
  - 实体通过 NER 从文本和数据中抽取
  - 关系连接实体，可通过基于模式、基于模型或基于 LLM 的方法抽取

- **Embedding**
  - 编码文本或图结构语义含义的数值向量
  - 用于语义搜索、聚类和基于相似度的检索

- **GraphRAG**
  - 将向量搜索与图遍历相结合
  - 同时利用 Embedding 和图结构，检索内容丰富、上下文感知的信息

- **本体（Ontology）**
  - 对某个领域中的类、关系和约束的形式化模型
  - 用于统一语义、支持推理、集成异构数据

- **质量与治理**
  - 质量指标（完整性、一致性、准确性、覆盖率）
  - 知识图谱层面的冲突检测与消解

````python
import os
from pathlib import Path

os.environ["SEMANTICA_API_KEY"] = "your_openai_key"
os.environ["SEMANTICA_EMBEDDING_PROVIDER"] = "openai"
os.environ["SEMANTICA_MODEL_NAME"] = "gpt-4"

config_text = """api_keys:
  openai: your_key_here
  anthropic: your_key_here
embedding:
  provider: openai
  model: text-embedding-3-large
  dimensions: 3072
knowledge_graph:
  backend: networkx
  temporal: true
"""
Path("config.yaml").write_text(config_text, encoding="utf-8")
Path("config.yaml").read_text(encoding="utf-8")
````

## 最小端到端流水线

下一个示例展示如何按顺序显式使用多个模块。这与前面讨论的架构一一对应：

1. 接入一个文档目录
2. 将其解析为结构化文档
3. 规范化文本
4. 抽取实体和关系
5. 构建并分析知识图谱
6. 生成 Embedding 并存入向量存储
7. 执行一次混合语义搜索查询

````python
from semantica.ingest import FileIngestor
from semantica.parse import DocumentParser
from semantica.normalize import TextNormalizer
from semantica.semantic_extract import NERExtractor, RelationExtractor
from semantica.kg import GraphBuilder, GraphAnalyzer
from semantica.embeddings import EmbeddingGenerator
from semantica.vector_store import VectorStore, HybridSearch

ingestor = FileIngestor()
documents = ingestor.ingest(str(docs_dir))

parser = DocumentParser()
parsed_docs = parser.parse(documents)

normalizer = TextNormalizer()
normalized_docs = normalizer.normalize(parsed_docs)

ner = NERExtractor()
entities = ner.extract(normalized_docs)
rel_extractor = RelationExtractor()
relationships = rel_extractor.extract(normalized_docs, entities)

builder = GraphBuilder()
kg = builder.build(entities, relationships)
analyzer = GraphAnalyzer()
metrics = analyzer.analyze(kg)

emb_generator = EmbeddingGenerator()
embeddings = emb_generator.generate_embeddings(documents, data_type="text")

vec_store = VectorStore()
vec_store.store(embeddings, documents, metadata={})
hybrid = HybridSearch(vec_store)
search_results = hybrid.search("Apple founders", top_k=3)
len(search_results)
````

## 可视化

Semantica 内置了强大的可视化模块。这里我们用上面构建的知识图谱创建一个交互式网络图。

````python
from semantica.visualization import KGVisualizer

# Create a visualizer instance
viz = KGVisualizer(layout="force", color_scheme="vibrant")

# Generate an interactive network visualization
# This returns a Plotly figure object that renders in the notebook
fig = viz.visualize_network(kg, output="interactive")
fig.show()
````

## 本体生成

你还可以从抽取出的实体和关系中自动生成本体（即所在领域的形式化模型）。

````python
from semantica.ontology import OntologyGenerator

generator = OntologyGenerator(base_uri="https://example.org/ontology/")

# Generate ontology from the extracted data
ontology = generator.generate_ontology({
    "entities": entities,
    "relationships": relationships
})

# View inferred classes
[cls["name"] for cls in ontology.get("classes", [])[:5]]
````

## 进阶：文本切分与分块

对 RAG 应用来说，把文档切分成更小的块至关重要。Semantica 为此提供了 `split` 模块。

````python
from semantica.split import TextSplitter

splitter = TextSplitter(chunk_size=100, chunk_overlap=20)
chunks = splitter.split_documents(documents)

print(f"Original documents: {len(documents)}")
print(f"Generated chunks: {len(chunks)}")
````

## 进阶：推理与推断

`reasoning` 模块允许你使用逻辑规则从已有知识中推导出新的事实。

````python
from semantica.reasoning import Reasoner

# Simple rule: If X founded Y, then X works_for Y
rule = """
IF (?x founded ?y) THEN (?x works_for ?y)
"""

reasoner = Reasoner()
reasoner.add_rule(rule)
inferred_facts = reasoner.infer_facts(kg)

print(f"Inferred {len(inferred_facts)} new facts")
````

## 进阶：导出与持久化

你可以将知识图谱保存到磁盘，或导出为 CSV、JSON、RDF 等标准格式。

````python
from semantica.export import GraphExporter

exporter = GraphExporter()
exporter.export(kg, format="json", output_path="knowledge_graph.json")
print("Graph exported to knowledge_graph.json")
````

## 使用核心 `Semantica` 类

对于更复杂的系统，你可以直接使用 `Semantica` 核心类和配置对象。这样可以使用生命周期管理、插件注册和编排辅助能力。

````python
from semantica.core import Semantica, ConfigManager

config_manager = ConfigManager()
config = config_manager.load_from_file("config.yaml")

framework = Semantica(config=config)
framework.initialize()

kb_result = framework.build_knowledge_base(
    sources=[str(docs_dir)],
    embeddings=True,
    graph=True,
)

framework.shutdown()
sorted(kb_result.keys())
````
